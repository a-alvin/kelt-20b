## KELT-20b Evolution

This notebook details the analytical model built to explore whether equilibrium tides alone could explain the current orbital architecture of the KELT-20 planetary system. KELT-20 is a young, A-type star with a closely orbiting gas giant companion&mdash;the hot Jupiter KELT-20b. The planet's orbit has a semi-major axis of 0.0542 Astronomical Units (AU) and an eccentricity of 0. The KELT-20 system has an estimated age constraint of 58&plusmn;5 Myr (Distler 2026). This may not have been enough time for the planet to migrate inward due to equilibrium tides. In this notebook, we explored the equilibrium tide model and found that the planet could **not** have migrated inward within the age of the system, indicating that another mechanism must be needed to explain this orbital architecture.

The equations used in the **Analytical Model** section are found in Hut (1981). The formatting and content of this notebook is greatly inspired by two REBOUNDx example notebooks: [TidesConstantTimeLag.ipynb](https://github.com/dtamayo/reboundx/blob/main/ipython_examples/TidesConstantTimeLag.ipynb) and [TidesSpinPseudoSynchronization.ipynb](https://github.com/dtamayo/reboundx/blob/main/ipython_examples/TidesSpinPseudoSynchronization.ipynb).

### Parameters

We used values from Lund (2017) for the stellar and planetary masses and radii and the stellar luminosity. The stellar Love number and the radius of gyration are found in Claret (2019) from astroseismic models. Vsin<em>i</em> comes from Rainer (2021). The parameters inputted into the model are in units of $M_\odot$, AU, and years for simplicity.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
np.random.seed(42)

# Stellar parameters - KELT-20
stellar_mass = 1.76  # solar masses
stellar_radius = 1.565 * 0.00465  # AU (1.565 solar radii -> AU)
k2_s = 0.0045  # stellar Love number
rg = 0.21  # radius of gyration

# Planetary parameters - KELT-20b
planetary_mass = 1 * 0.000954588  # solar masses; choose between 3.382 (max) and 1 (min)
planetary_radius = 1.741 * 0.000477895  # AU (1.741 Jupiter radii -> AU) 

G = 4*np.pi**2  # AU^3 / (yr^2 Msun)
```

#### Rotation Period

The rotation period of the star has great effects on the efficiency of tidal migration. As KELT-20 is an extremely young A-type star, it is expected to have a rapid rotation rate. To calculate the rotation period, we used the stellar radius and projected rotational velocity vsin<em>i</em>. According to Lund, the inclination of the star is constrained to 24&deg; < <em> i </em> < 156&deg;, so we calculated a maximum period (with 90&deg; inclination) and a minimum period (with 156&deg; inclination) to use when modeling the orbital evolution.

```python
vsini = 116.25  # km/s
r_km = 1.565 * 695700  # stellar radius in kilometers

p_upper = 2 * np.pi * r_km * np.sin(90 * np.pi/180) / vsini / 86400  # days
p_lower = 2 * np.pi * r_km * np.sin(156 * np.pi/180) / vsini / 86400  # days
print(f"Minimum period: {p_lower:.2} days,", f"Maximum period: {p_upper:.2f} days")

# Setting stellar rotation to be used in the model, choosing either p_upper or p_lower.
period =  p_lower # days
print(f"Chosen period: {period:.2} days")
period = period / 365  # years
omega_0 = 2 * np.pi / period # cycles per year
```

The minimum period that is determined in the previous code, 0.28 days, seems quite small. To check if this value is physically reasonable, we calculated the critical velocity, or the maximum rotational velocity the star can have before its gravity is unable to hold it together. Critical velocity is independent of inclination as it only depends on the relationship between the gravitational binding force and the centripetal force,
$$
\frac{GM_{\star}m}{R_\star^2}=\frac{mv^2}{R_{\star}} 
$$
$$
v=\bigg(\frac{GM_\star}{R_{\star}}\bigg)^{1/2}
$$


We used this critical velocity to calculate the critical period. This calculation does not include a sin<em>i</em> term as it is a maximum possible velocity, which would occur for <em>i</em> = 90&deg;, sin(90&deg;) = 1.

We found that the critical period was around 0.17 days, which indicates that the minimum period we calculated previously is physically reasonable.

We also defined the stellar parameters again in centimeter-gram-second units to ensure accuracy, and for later use in calculating the convective friction time.

```python
# Parameters in centimeter-gram-second units
G_cgs = 6.674e-8  # [cm^3 / g s^2]
m_cgs = stellar_mass * 1.989e33  # stellar mass in grams
r_cgs = 1.565 * 6.955e10  # stellar radius in cm 
l_cgs = 12.7 * 4e33  # stellar luminosity in erg/s

# Critical velocity
critical_v_cgs = np.sqrt(G_cgs * m_cgs / r_cgs)  # cm/s
print(f"Critical velocity: {(critical_v_cgs / 100000):.5} km/s")

# Critical period
critical_p_days = 2 * np.pi * r_cgs / critical_v_cgs / 86400  # days
print(f"Critical period: {critical_p_days:.2} days")
```

#### Initial Orbital Geometry

In analyzing the KELT-20 system, it is important to consider how far away the planet could have formed before migrating. Gas giants typically form beyond the snow line, or the distance from the star at which the temperature is low enough for volatiles (water, carbon dioxide, etc.) to be in their solid state.

The snow line of a bright star scales as $r=2.7 *(L_\star/L_\odot)^{1/2}$. Another relation is $r=2.7 *(M_\star/M_\odot)^2$. We find a snow line of 8.4&ndash;9.6 AU, depending on the relation, which is much further out than that of the Sun. This indicates that we should explore initial semi-major axis ($a$) values beyond the snow line. We explore initial $a$ values from 7&ndash;20 AU.

We calculate the eccentricity ($e$) of the initial orbit using a simple relation with the semi-major axis. The current orbit of the star has $a = 0.0542$ AU and $e = 0$. A planet with a highly eccentric orbit will circularize to a radius of approximately twice its initial periastron distance ($q)$, or distance of closest approach. This indicates that the planet must have begun with an initial distance of &asymp;0.025 AU at periastron. With a known periastron distance and semi-major axis, calculating the necessary eccentricity can be easily done with $e = 1-q/a$.

```python
# Final orbital conditions
a_final = 0.0542
e_final = 0

# Snow line
rsnow=2.7 *np.sqrt(12.7)  # 12.7 = Lstar/Lsun
print(f"Snow line 1: {rsnow:.2} AU")

rsnow=2.7 *(1.76)**2  # 1.76 = Mstar/Msun
print(f"Snow line 2: {rsnow:.2} AU")

# Setting initial orbital geometry
a_array = np.linspace(7, 20, 20)
e_array = 1 - 0.025 / a_array
```

#### Constant Time Lag Parameter

We modeled equilibrium tides with the constant time lag model described in Hut (1981). We used the definitions of convective friction time and constant time lag as outlined in the REBOUNDx example notebook [TidesConstantTimeLag.ipynb](https://github.com/dtamayo/reboundx/blob/main/ipython_examples/TidesConstantTimeLag.ipynb), which draws from Zahn (1989) and Baronett et al. (2022). The constant time lag parameter ($\tau$) describes how much energy is dissipated in the envelope of the star. $\tau$ is defined as
$$
\tau = \frac{2R_{\star}^3}{GM_{\star}t_f}
$$
where $R_{\star}$ and $M_{\star}$ are the stellar mass and radius, $G$ is the gravitational constant, and $t_f$ is the convective friction time, 
$$
t_f = \bigg(\frac{M_{\star}R_{\star}^2}{L_{\star}}\bigg)^{1/3},
$$
where $L_{\star}$ is the stellar luminosity. These parameters are calculated below in units of years.

```python
# Convective friction time
t_f_s = (m_cgs * r_cgs **2 / l_cgs)**(1/3)  # seconds
t_f = t_f_s / 31556952  # years
print(f"Convective friction time: {t_f:.4} years")

# Tidal time lag
tau = 2 * stellar_radius**3 / (G * stellar_mass * t_f) #CHANGE BACK t_r
print(f"Constant time lag: {tau:.2} years")
```

### Hut Model

We used the differential equations defined in Hut (1981) to model the evolution of the semi-major axis and eccentricity of the planet's orbit as well as the rotation rate of the star. We added a condition to effectively stop recording change if the semi-major axis decreased to below the radius of the star to prevent recording nonphysical results. The hut_model function is defined similarly to the model in the [TidesSpinPseudoSynchronization.ipynb](https://github.com/dtamayo/reboundx/blob/main/ipython_examples/TidesSpinPseudoSynchronization.ipynb) REBOUNDx example notebook. 

```python
def f1_hut(e2):
    return 1 + (31/2)*e2 + (255/8)*e2**2 + (185/16)*e2**3 + (25/64)*e2**4

def f2_hut(e2):
    return 1 + (15/2)*e2 + (45/8)*e2**2 + (5/16)*e2**3

def f3_hut(e2):
    return 1 + (15/4)*e2 + (15/8)*e2**2 + (5/64)*e2**3

def f4_hut(e2):
    return 1 + (3/2)*e2 + (1/8)*e2**2

def f5_hut(e2):
    return 1 + 3*e2 + (3/8)*e2**2


def hut_model(params, t, mstar, rstar, k2star, mplanet, tau, rg):
    a, e, omega = params
    
    if a < rstar:
        return [0,0,0]
    
    k =  k2star
    
    n = np.sqrt(G*(mstar + mplanet)/a**3)

    T = rstar**3 / (G * mstar *  tau) 

    q = mplanet / mstar

    rg = rg
    
    da_dt = -6 * (k / T) * q * (1 + q) * (rstar / a)**8 * a * (1 - e**2)**(-15/2) * (
         f1_hut(e**2) - (1 - e**2)**(3/2) * f2_hut(e**2) * omega / n)

    de_dt = -27 * (k / T) * q * (1 + q) * (rstar / a)**8 * e * (1 - e**2)**(-13/2) * (
        f3_hut(e**2) - 11/18 * (1 - e**2)**(3/2) * f4_hut(e**2) * omega / n)

    domega_dt = 3 * (k / T) * (q / rg)**2 * (rstar / a)**6 * n * (1-e**2)**(-6) * (
        f2_hut(e**2) - (1 - e**2)**(3/2) * f5_hut(e**2) * omega / n)

    
    return [da_dt, de_dt, domega_dt]
```

### Running Model and Plotting

We define a time array to best show the result of the plot. We evaluate the analytical model for each initial semi-major axis and eccentricity pair using the scipy odeint differential equation integrator, then plot the evolution of the semi-major axis, eccentricity, and angular frequency versus time. 

```python
# Time
tmax = 3e9
Nout = 2000
hut_times = np.linspace(0, tmax, Nout)

print(f"Mp: {planetary_mass/0.000954588} MJ, Prot: {period*365:.2} days")

# Setting up plot
fig, ax = plt.subplots(3, 1, figsize = (4,10), sharex=True)
cmap = plt.get_cmap("viridis_r")
colors = cmap(np.linspace(0, 1, 20))

for i in range(len(a_array)):
    a0 = a_array[i]
    e0 = e_array[i]
    
    initial_cond = [a0, e0, omega_0] 
    
    hut_solution = odeint(hut_model, initial_cond, hut_times, 
                               args = (stellar_mass, stellar_radius, k2_s, 
                                      planetary_mass, tau, rg))
    
    a_sol = hut_solution[:, 0]
    e_sol = hut_solution[:, 1]
    o_sol = hut_solution[:, 2]
    
    ax[0].plot(hut_times, a_sol, c = colors[i])
    ax[1].plot(hut_times, e_sol, c = colors[i])
    ax[2].plot(hut_times, o_sol, c = colors[i])

# One set_title line can be uncommented for each chosen mass/period combination.
# ax[0].set_title(r'Maximum $M_P$, Maximum $P_{rot}$')
# ax[0].set_title(r'Minimum $M_P$, Maximum $P_{rot}$')
ax[0].set_title(r'Minimum $M_P$, Minimum $P_{rot}$')
# ax[0].set_title(r'Maximum $M_P$, Minimum $P_{rot}$')

# Semi-major axis subplot
ax[0].set_ylabel("Semi-Major Axis (AU)", fontsize='large')
ax[0].set_xscale("log")
ax[0].set_xlim(4e7, tmax)
ax[0].axvspan(53e6, 63e6, alpha = 0.7, color = "gray")  # bar spanning KELT-20 age for comparison

# Eccentricity subplot
ax[1].set_ylabel("Eccentricity", fontsize='large')
ax[1].set_xscale("log")
ax[1].set_xlim(4e7, tmax)
ax[1].axvspan(52e6, 64e6, alpha = 0.7, color = "gray")

# Angular frequency subplot
ax[2].set_ylabel("Stellar Angular Frequency \n (cycles/yr)", fontsize='large')
ax[2].set_xlabel("Time (yr)", fontsize='large')
ax[2].set_xscale("log")
ax[2].set_xlim(4e7, tmax)
ax[2].axvspan(52e6, 64e6, alpha = 0.7, color = "gray", label = "KELT-20 age range")

## Adding text to plot with values of mass and P_rot depends on which plot we choose.

# max max, min max, min min
ax[0].set_ylim(6,26)
ax[0].text(0.7*10**8, 24, r"$M_{\mathrm{P}}=$ "+ f"{planetary_mass/0.000954588}"+ r" $M_{\mathrm{J}}$", fontsize=11)
ax[0].text(0.7*10**8, 22, r"$P_{\mathrm{rot}}=$ " +f"{period*365:.2} days", fontsize=11)

# max min
# ax[0].text(0.7*10**8, 42, r"$M_{\mathrm{P}}=$ "+ f"{planetary_mass/0.000954588}"+ r" $M_{\mathrm{J}}$", fontsize=11)
# ax[0].text(0.7*10**8, 38, r"$P_{\mathrm{rot}}=$ " +f"{period*365:.2} days", fontsize=11)

ax[2].legend(loc = "lower center")

## Option to add outline around figures
# fig.tight_layout()
# fig.patch.set_linewidth(1)
# fig.patch.set_edgecolor('black')

## Option to save or show the figure (must change the filename)
# plt.savefig("max_min.png", dpi=200, format="png", facecolor="white", pad_inches=0.2, bbox_inches = "tight") #, edgecolor=fig.get_edgecolor())
plt.show()
```

## Conclusion

We find that high-eccentricity migration due to equilibrium tides alone cannot recreate the KELT-20 system within its age constraint of 58 Myr, and in fact the planet will migrate outward due to the rapid rotation of the A-type host.
